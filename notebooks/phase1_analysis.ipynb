# Phase 1 Analysis: Visual Inspection & Failure Analysis
This notebook provides tools to visually inspect the synthetic tasks generated during Phase 1, specifically focusing on the tasks where the Heuristic baseline failed.

```python
import os
import json
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

sys.path.insert(0, os.path.dirname(os.path.abspath('')))
from core.task import ArcTask
```

### ARC Colors Configuration

```python
ARC_COLORS = [
    '#000000', # 0: black (background)
    '#0074D9', # 1: blue
    '#FF4136', # 2: red
    '#2ECC40', # 3: green
    '#FFDC00', # 4: yellow
    '#AAAAAA', # 5: grey
    '#F012BE', # 6: fuchsia / magenta
    '#FF851B', # 7: orange
    '#7FDBFF', # 8: teal / cyan
    '#870C25'  # 9: brown
]
cmap = mcolors.ListedColormap(ARC_COLORS)
norm = mcolors.Normalize(vmin=0, vmax=9)

def plot_grid(ax, grid, title=""):
    ax.imshow(grid, cmap=cmap, norm=norm)
    ax.grid(True, which='both', color='lightgrey', linewidth=0.5)
    ax.set_xticks([x - 0.5 for x in range(1 + grid.shape[1])])
    ax.set_yticks([y - 0.5 for y in range(1 + grid.shape[0])])
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.set_title(title)

def visualize_task(task):
    n_train = len(task.train_pairs)
    n_test = len(task.test_pairs)
    fig, axes = plt.subplots(n_train + n_test, 2, figsize=(6, 3 * (n_train + n_test)))
    if n_train + n_test == 1:
        axes = [axes]
    
    for i, pair in enumerate(task.train_pairs):
        plot_grid(axes[i][0], pair['input'], f"Train {i+1} Input")
        plot_grid(axes[i][1], pair['output'], f"Train {i+1} Output")
        
    for i, pair in enumerate(task.test_pairs):
        plot_grid(axes[n_train + i][0], pair['input'], f"Test {i+1} Input")
        plot_grid(axes[n_train + i][1], pair['output'], f"Test {i+1} Output")
        
    plt.suptitle(f"Task: {task.task_id} (Rule: {task.rule_type})\n", fontsize=14)
    plt.tight_layout()
    plt.show()
```

### Load Synthetic Tasks

```python
data_dir = Path("../data/synthetic/held_out")
tasks = []
for root, _, files in os.walk(data_dir):
    for f in files:
        if f.endswith('.json'):
            tasks.append(ArcTask.load(Path(root) / f))
```

```python
# Visualize a random task
import random
random_task = random.choice(tasks)
visualize_task(random_task)
```
